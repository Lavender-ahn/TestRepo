## Extract information from the given web site
You will extract the data from the below web site: <br>


```python
#this url contains the data you need to scrape
url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DA0321EN-SkillsNetwork/labs/datasets/Programming_Languages.html"
```

The data you need to scrape is the **name of the programming language** and **average annual salary**.<br> It is a good idea to open the url in your web broswer and study the contents of the web page before you start to scrape.


Import the required libraries


```python
# Your code here
import pandas as pd
import warnings
warnings.simplefilter("ignore")
from bs4 import BeautifulSoup
import requests
```

Download the webpage at the url


```python
#your code goes here
html = requests.get(url).text
```

Create a soup object


```python
#your code goes here
data_bs = BeautifulSoup(html, 'html.parser')
```

```python
print(data_bs.prettify())
```

Scrape the `Language name` and `annual average salary`.


```python
#your code goes here
table = pd.DataFrame(columns=['Language', 'Annual Average Salary'])

for row in data_bs.find_all('tr'):
    cols = row.find_all('td')
    if cols:
      language = cols[1].string
      annual_average = cols[3].string

    new_row = pd.DataFrame([{'Language': language, 'Annual Average Salary': annual_average}])

    table = pd.concat([table, new_row], ignore_index=True)

table
```

```python
table = table.drop(table.index[0])
table.reset_index(drop=True)

table
```

Save the scrapped data into a file named *popular-languages.csv*


```python
# your code goes here
table.to_csv('popular-languages.csv', index=False)
```
